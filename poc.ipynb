```python
import os
from google import genai
from dotenv import load_dotenv
load_dotenv()
client=genai.Client(api_key=os.getenv("GEMINI_API_KEY"))
def add_use_chat(messages,text):
    messages.append({"role": "user", "parts": [{"text": text}]})



def ai_answer(messages,text):
    messages.append({"role": "model", "parts": [{"text": text}]})
def chat_model(messages):
    response=client.models.generate_content(
        model="gemini-3.1-flash-lite",
        contents=messages
    )
    return response.text
messages=[]
# add_use_chat(messages,"What is Quantum Computing?,answer in one sentence.")
# answer=chat_model(messages)
# print(answer)
# ai_answer(messages,answer)
# add_use_chat(messages,"explain one example")
# answer=chat_model(messages)
# ai_answer(messages,answer)
for i in range(10):
    text=str(input("Enter a query:"))
    print(text)
    add_use_chat(messages,text)
    answer=chat_model(messages)
    print("Answer:",answer)
    ai_answer(messages,answer)
    
    
```
